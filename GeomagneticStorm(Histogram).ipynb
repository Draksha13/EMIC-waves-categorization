```python
import pandas as pd
from datetime import datetime

# 1. Define the storm categorization logic
def categorize_storm(val):
    if val <= -300: return "Superintense"
    if val <= -150: return "Intense"
    if val <= -80:  return "Moderate"
    if val <= -50:  return "Weak"
    return "Quiet/Sub-storm"

# 2. Define the exact columns based on your file structure
cols = [
    'm_yr', 'm_mo', 'm_da', 'm_hr', 'm_mn', 'm_sc', 'm_sym',
    'i_yr', 'i_mo', 'i_da', 'i_hr', 'i_mn', 'i_sc', 'i_sym',
    'storm_no', 'ips_sym'
]

# 3. Load and process the file
try:
    # Read the file (skipping the 2-line header) [cite: 149]
    df = pd.read_csv('/content/SYM-H_storms_1981_2025.txt', sep='\s+', skiprows=2, names=cols)

    processed_list = []

    for _, row in df.iterrows():
        # Create timestamps for Onset and Peak
        try:
            mpo_time = datetime(int(row['m_yr']), int(row['m_mo']), int(row['m_da']), int(row['m_hr']), int(row['m_mn']))
            int_time = datetime(int(row['i_yr']), int(row['i_mo']), int(row['i_da']), int(row['i_hr']), int(row['i_mn']))

            # Duration = Peak Time - Onset Time
            main_phase_dur = (int_time - mpo_time).total_seconds() / 3600

            processed_list.append({
                'Storm_No': int(row['storm_no']),
                'Main_Phase_Onset': mpo_time,
                'Recovery_Phase_Initiation': int_time,
                'Main_Phase_Duration_Hrs': round(main_phase_dur, 2),
                'Min_SYM_H': row['i_sym'],
                'Category': categorize_storm(row['i_sym']),
                'Year': mpo_time.year
            })
        except ValueError:
            continue # Skip rows with invalid date data

    # 4. Save to a CSV file on your PC
    final_df = pd.DataFrame(processed_list)
    output_filename = 'Processed_Storm_Data.csv'
    final_df.to_csv(output_filename, index=False)

    print(f"Success! Data saved to {output_filename}")
    print(final_df.head())

except FileNotFoundError:
    print("Error: 'SYM-H_storms_1981_2025.txt' not found in the current directory.")
```

```python
import pandas as pd
import pyspedas
from datetime import datetime, timedelta
import numpy as np

storm_data_list = [
    [1981, 2, 6, 12, 59, 1981, 2, 7, 1, 7],
    [1981, 3, 5, 6, 5, 1981, 3, 5, 17, 48],
    [1981, 4, 12, 23, 20, 1981, 4, 13, 7, 9],
    [1981, 5, 10, 22, 13, 1981, 5, 11, 3, 49]
]

results = []

print("Connecting to OMNI database...")

for s in storm_data_list:
    try:
        start = datetime(s[0], s[1], s[2], s[3], s[4])
        end = datetime(s[5], s[6], s[7], s[8], s[9])

        trange = [
            (start - timedelta(hours=1)).strftime('%Y-%m-%d/%H:%M:%S'),
            (end + timedelta(hours=1)).strftime('%Y-%m-%d/%H:%M:%S')
        ]

        # 🔹 Directly get dictionary (no pytplot)
        data = pyspedas.omni.data(
            trange=trange,
            datatype='1min',
            notplot=True,
            no_update=True
        )

        # Extract safely
        bz = data.get('bz_gsm')
        pdyn = data.get('p_dyn')

        if bz is not None and pdyn is not None:
            # bz[1] = values, bz[0] = time
            bz_values = bz['y'] if isinstance(bz, dict) else bz[1]
            pdyn_values = pdyn['y'] if isinstance(pdyn, dict) else pdyn[1]

            # Convert to numpy arrays (important)
            bz_values = np.array(bz_values)
            pdyn_values = np.array(pdyn_values)

            # Remove fill values (very important for OMNI)
            bz_values = bz_values[(bz_values > -1e4) & (bz_values < 1e4)]
            pdyn_values = pdyn_values[(pdyn_values > 0) & (pdyn_values < 100)]

            min_bz = np.min(bz_values) if len(bz_values) > 0 else np.nan
            max_pdyn = np.max(pdyn_values) if len(pdyn_values) > 0 else np.nan
        else:
            min_bz, max_pdyn = np.nan, np.nan

        results.append({
            'MPO_Start': start,
            'Intensity_Peak': end,
            'Min_Bz_GSM': min_bz,
            'Max_Pdyn': max_pdyn
        })

        print(f"Processed: {start.date()} | Bz: {min_bz} | Pdyn: {max_pdyn}")

    except Exception as e:
        print(f"Error on date {s[0]}-{s[1]}-{s[2]}: {e}")

df_final = pd.DataFrame(results)

print("\n--- Final Extracted Data ---")
print(df_final)
```

```python
import pandas as pd
from datetime import datetime
import matplotlib.pyplot as plt
from matplotlib.backends.backend_pdf import PdfPages
import os

# 1. Define the storm categorization logic
def categorize_storm(val):
    if val <= -300: return "Superintense"
    if val <= -100: return "Intense"
    if val <= -80:  return "Moderate"
    if val <= -50:  return "Weak"
    return "Quiet/Sub-storm"

# 2. Updated columns to include Bz and Pdyn (assuming they follow ips_sym or similar)
# Ensure your .txt file structure matches this order!
cols = [
    'm_yr', 'm_mo', 'm_da', 'm_hr', 'm_mn', 'm_sc', 'm_sym',
    'i_yr', 'i_mo', 'i_da', 'i_hr', 'i_mn', 'i_sc', 'i_sym',
    'storm_no', 'ips_sym', 'bz_val', 'pdyn_val'
]

try:
    # 3. Load the file
    df = pd.read_csv('SYM-H_storms_1981_2025.txt', sep='\s+', skiprows=2, names=cols)
    processed_list = []

    for _, row in df.iterrows():
        try:
            # Create timestamps
            mpo_time = datetime(int(row['m_yr']), int(row['m_mo']), int(row['m_da']), int(row['m_hr']), int(row['m_mn']))
            int_time = datetime(int(row['i_yr']), int(row['i_mo']), int(row['i_da']), int(row['i_hr']), int(row['i_mn']))

            # Duration Calculation
            main_phase_dur = (int_time - mpo_time).total_seconds() / 3600

            processed_list.append({
                'Storm_No': int(row['storm_no']),
                'Onset': mpo_time.strftime('%y-%m-%d %H:%M'),
                'Peak': int_time.strftime('%y-%m-%d %H:%M'),
                'Dur_Hrs': round(main_phase_dur, 1),
                'SYM_H': row['i_sym'],
                'Bz': row['bz_val'],      # <--- New Column
                'Pdyn': row['pdyn_val'],  # <--- New Column
                'Category': categorize_storm(row['i_sym'])
            })
        except ValueError:
            continue

    final_df = pd.DataFrame(processed_list)

    # --- PDF EXPORT LOGIC ---

    downloads_path = os.path.join(os.path.expanduser("~"), "Downloads")
    pdf_filename = os.path.join(downloads_path, 'Storm_Data_Full.pdf')

    # Use a wider figure to accommodate more columns
    fig, ax = plt.subplots(figsize=(14, 10))
    ax.axis('tight')
    ax.axis('off')

    # We display the top 60 rows to prevent the PDF from becoming unreadable
    # If you need ALL rows, you'd typically loop and create multiple pages
    display_df = final_df.head(60)

    table = ax.table(
        cellText=display_df.values,
        colLabels=display_df.columns,
        loc='center',
        cellLoc='center'
    )

    table.auto_set_font_size(False)
    table.set_fontsize(7) # Slightly smaller font for more columns
    table.scale(1.1, 1.5)

    with PdfPages(pdf_filename) as pdf:
        pdf.savefig(fig, bbox_inches='tight')

    plt.close()

    print(f"Success! PDF with Bz and Pdyn saved to: {pdf_filename}")

except FileNotFoundError:
    print("Error: 'SYM-H_storms_1981_2025.txt' not found.")
except KeyError as e:
    print(f"Error: Column {e} not found. Check if your text file has Bz and Pdyn columns.")
```

```python
import pandas as pd
from fpdf import FPDF
from google.colab import files
import os

# --- (Your existing processing logic here to create final_df) ---
# Assuming final_df has columns: Storm_No, Onset, Peak, Dur_Hrs, SYM_H, Bz, Pdyn, Category

class PDF(FPDF):
    def header(self):
        self.set_font('Arial', 'B', 12)
        self.cell(0, 10, 'Geomagnetic Storm Analysis Report (1981-2025)', 0, 1, 'C')
        self.ln(5)

    def footer(self):
        self.set_y(-15)
        self.set_font('Arial', 'I', 8)
        self.cell(0, 10, f'Page {self.page_no()}', 0, 0, 'C')

def export_to_pdf(df, filename="Storm_Data.pdf"):
    pdf = PDF(orientation='L') # Landscape to fit Bz and Pdyn
    pdf.add_page()
    pdf.set_font("Arial", size=8)

    # Table Header
    col_width = pdf.w / (len(df.columns) + 1)
    pdf.set_fill_color(200, 220, 255)
    for col in df.columns:
        pdf.cell(col_width, 10, str(col), border=1, fill=True, align='C')
    pdf.ln()

    # Table Rows
    pdf.set_fill_color(255, 255, 255)
    for i, row in df.iterrows():
        for item in row:
            pdf.cell(col_width, 8, str(item), border=1, align='C')
        pdf.ln()

    pdf.output(filename)
    return filename

# Generate and Download
pdf_file = export_to_pdf(final_df)
final_df.to_csv("Storm_Data.csv", index=False)

# This triggers the browser download automatically in Colab
files.download(pdf_file)
files.download("Storm_Data.csv")
```
