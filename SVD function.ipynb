```python
import numpy as np

# Example covariance matrix
M = np.array([[2, 1, 0],
              [1, 2, 1],
              [0, 1, 2]])
print("Original M:\n", M)
```

```python
# Step 1: Compute Eigenvalues and Eigenvectors
eig_vals, eig_vecs = np.linalg.eig(M)
print("Computed U (Left Singular Vectors):\n", U)
print("Computed S (Singular Values Matrix):\n", S)
```

```python
# Step 2: Sort by descending eigenvalues
sorted_indices = np.argsort(eig_vals)[::-1]
eig_vals_sorted = eig_vals[sorted_indices]
eig_vecs_sorted = eig_vecs[:, sorted_indices]
eig_vecs_sorted
```

```python
eig_vals_sorted
```

```python
# Step 3: Compute Singular Values (Square Root of Eigenvalues)
singular_values = np.sqrt(eig_vals_sorted)
singular_values
```

```python
# Step 4: Construct SVD Matrices
U = eig_vecs   # Left singular vectors
U
```

```python
S = np.diag(singular_values)  # Singular values matrix
S
```

```python
# Since M is symmetric, V = U
V_T = U.T  # Right singular vectors (transpose of U)
V_T
print("Computed V_T (Right Singular Vectors Transposed):\n", V_T)
```

```python
# Step 5: Verify that M ≈ U * S * V_T
M_reconstructed = U @ S @ V_T
M_reconstructed
```

```python



print("Reconstructed M (U S V_T):\n", M_reconstructed)
```

```python
import numpy as np

# Example eigenvector matrix (U)
U = np.array([
    [-0.5,  0.707,  0.5],
    [-0.707, 4.05e-16, -0.707],
    [-0.5, -0.707,  0.5]
])

# Example eigenvalues (unsorted)
eig_vals = np.array([2, 1, 3])  # Example eigenvalues (not sorted)

# Step 1: Sort indices by descending eigenvalues
sorted_indices = np.argsort(eig_vals)[::-1]

# Step 2: Sort eigenvalues and reorder eigenvectors
eig_vals_sorted = eig_vals[sorted_indices]
U_sorted = U[:, sorted_indices]

# Print results
print("Sorted Eigenvalues:\n", eig_vals_sorted)
print("Sorted U (Eigenvectors Corresponding to Sorted Eigenvalues):\n", U_sorted)
```
